```python
tips.groupby('time').agg(
    bills=('total_bill','size'),
    avg_bill=('total_bill','mean'),
    avg_party=('size','mean')
)
```

```python
smry = tips.groupby('day').agg(
    bills=('total_bill','size'),
    avg_bill=('total_bill','mean'),
    avg_tip=('tip','mean')
).round(2)
smry
```

```python
print(smry.columns.tolist())
print(smry.index.to_list())
flat=smry.reset_index()
print(flat.columns.tolist())
print(flat[flat["day"]=="Sun"])
flat.head()
```
