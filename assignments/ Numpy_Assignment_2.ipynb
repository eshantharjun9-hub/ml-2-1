```python
# Award medals to the scores below using nested np.where: 'Gold' for 90 and above, 'Silver' for 75 and above, 'Bronze' for 60 and above, and 'None' otherwise. Then count how many students received each medal.

scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])
gold=scores[np.where(scores>90)]
silver=scores[np.where((scores>75) & (scores<90))]
bronze=scores[np.where((scores>60) & (scores<75))]
print(len(gold))
print(len(silver))
print(len(bronze))
```

```python
# Build a complete multi-step pipeline. Starting from the two arrays below: convert them into column vectors, join them into a single two-column table, attach a third column holding each row's total, then multiply the three-column table by w = [0.5, 0.2, 0.3] to obtain one score per row. Finally use np.where to mark each score as 1 if it is above the average score and 0 otherwise. Report the shape of every intermediate array.

import numpy as np

hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])

# 1. Convert to column vectors
hours_col = hours.reshape(-1, 1)
attendance_col = attendance.reshape(-1, 1)

print("hours_col:")
print(hours_col)
print("Shape:", hours_col.shape)

print("attendance_col:")
print(attendance_col)
print("Shape:", attendance_col.shape)


# 2. Join into a two-column table
cv = np.hstack((hours_col, attendance_col))

print("\nTwo-column table:")
print(cv)
print("Shape:", cv.shape)


# 3. Calculate total for each row
totals = cv.sum(axis=1).reshape(-1, 1)

print("\nTotals:")
print(totals)
print("Shape:", totals.shape)


# 4. Attach totals as third column
table = np.hstack((cv, totals))

print("\nThree-column table:")
print(table)
print("Shape:", table.shape)


# 5. Multiply by weights
w = np.array([0.5, 0.2, 0.3])

scores = table @ w

print("\nScores:")
print(scores)
print("Shape:", scores.shape)


# 6. Calculate average score
average = scores.mean()

print("\nAverage score:", average)
print("Shape:", np.array(average).shape)


# 7. Mark scores above average as 1, otherwise 0
result = np.where(scores > average, 1, 0)

print("\nFinal result:")
print(result)
print("Shape:", result.shape)
```
