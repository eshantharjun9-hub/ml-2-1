```python
str=input()
arr=[]
for file in str.split(','):
    file = file.strip()
    e=file.split('.')[-1].upper()
    if e not in arr:
        arr.append(e)
print(arr)
```

```python
str=input()
arr=[]
for ch in str:
    if ch.isdigit()==True:
        arr.append(ch)
if len(arr)!=10:
    print("Invalid")
else:
    num="".join(arr)
    print(f"({num[:3]}) {num[3:6]}-{num[6:]}")
```
