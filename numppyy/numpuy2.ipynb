```python
import numpy as np
```

```python
x=np.array([1,2.4,"hello"])
print(x)
```

```python
a1=np.arange(10)
print(a1)
a1=a1+5 ## add 5 to every element
print(a1)
a1=a1*5 ## multiply every element
print(a1)

x=np.arange(10)
y=np.arange(11,21)

print(x)
print(y)
print(x+y)
```

```python
x=np.arange(6).reshape(2,3)
y=np.arange(3)
print(x)
print(y)
print(x+y)
```

```python
x=np.arange(6).reshape(3,2)
y=np.arange(6).reshape(2,3)
print(x)
print(y)
print(x+y)
```

```python
x=np.arange(5)
y=np.arange(1,6)
print(x)
print(y)
print(x==y)
print(x>y)
print(x<y)
```
