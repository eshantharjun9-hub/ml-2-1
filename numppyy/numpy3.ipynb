```python
## vstack coloms must be equal
## hstack rows must be equal


a1=np.array([[1,3,4],[2,3,5]])
a2=np.array([[5,6,7],[8,9,10]])
a4=np.concatenate((a1,a2),axis=1)
a3=np.hstack((a1,a2))
a3=np.vstack((a1,a2))
print(a4)
# print(a3)
```

## axis=0 then vstack means runs to coloums
## axis=1 then hstack means runs to rowss
